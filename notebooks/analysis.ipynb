# Cinnamon cinnamaldehyde evidence inventory

The available source is a narrative note without sample-level data or adequate citations. The notebook preserves that distinction.

```python
import json
from pathlib import Path
claims = json.loads(Path("data/source_claims.json").read_text())
claims
```

```python
for item in claims["claims"]:
    lo, hi = item["range_percent"]
    if item["basis"] == "percent by weight":
        print(f"{item["material"]}: {lo:g}–{hi:g}% w/w = {lo*10:g}–{hi*10:g} mg/g (source-reported, unverified)")
    else:
        print(f"{item["material"]}: {lo:g}–{hi:g}% of essential oil (source-reported, unverified)")
```
